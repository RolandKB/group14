## Inspect raw data
Check a familiar municipality and look for municipalities with more than one price area.

```python
raw_path = DEFAULT_ROOT / "response.json"
with raw_path.open(encoding="utf-8") as file:
    payload = json.load(file)

municipalities = payload["kommuner"]
print("Municipalities:", len(municipalities))
for code in ("0301", "1103", "3301"):
    match = next((m for m in municipalities if m["number"] == code), None)
    print(code, match["name"] if match else "Not found",
          match["priceAreas"] if match else "")

multi_area = [m for m in municipalities if len(m["priceAreas"]) > 1]
print("Multiple price areas:", len(multi_area))
print([(m["number"], m["name"], m["priceAreas"]) for m in multi_area])
```
