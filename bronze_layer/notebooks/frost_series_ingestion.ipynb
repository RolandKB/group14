```python
from bronze_layer.frost_series import ingest
from dotenv import load_dotenv

load_dotenv()

result = ingest(
    start="2024-01-01",
    end="2026-01-01",
    station_ids=["SN40", "SN18700", "SN50540", "SN100", "SN10055"],
)

print("Checked:", result["station_count"])
print("Eligible:", result["eligible_count"])
print("Failed:", result["failed_count"])
```

## Inspect results

```python
for station, info in result["results"].items():
    print(station, info["status"], "candidate series:", info["candidate_count"])
    for candidate in info["candidates"][:3]:
        print(
            "  ",
            candidate["source_id"],
            candidate["resolution"],
            candidate["valid_from"],
            candidate["valid_to"],
        )
```
